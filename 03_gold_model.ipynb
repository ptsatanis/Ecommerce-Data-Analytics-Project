```python
%sql
CREATE OR REPLACE TABLE gold.dim_product AS
SELECT
    product_id,
    product_category_name,
    product_name_length,
    product_description_length,
    product_photos_qty,
    product_weight_g,
    product_length_cm,
    product_height_cm,
    product_width_cm
FROM silver.products;

SELECT COUNT(*) FROM gold.dim_product;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_seller AS
SELECT
    seller_id,
    seller_zip_code_prefix,
    seller_city,
    seller_state
FROM silver.sellers;

SELECT *
FROM gold.dim_seller
LIMIT 10;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_payments AS
SELECT
    order_id,
    payment_sequential,
    payment_type,
    payment_installments,
    payment_value
FROM silver.payments;

SELECT *
FROM gold.dim_payments
LIMIT 10;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_order_items AS
SELECT 
    order_id,
    order_item_id,
    product_id,
    shipping_limit_date,
    price,
    freight_value
FROM silver.order_items;

SELECT * 
FROM gold.dim_order_items
LIMIT 10;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_orders AS
SELECT
    order_id,
    customer_id,
    order_status,
    order_purchase_timestamp,
    order_approved_at,
    order_delivered_carrier_date,
    order_delivered_customer_date,
    order_estimated_delivery_date
FROM silver.orders;

SELECT *
FROM gold.dim_orders
LIMIT 10;
```

```python
%sql
SELECT COUNT(*) FROM gold.dim_customer;
SELECT COUNT(*) FROM gold.dim_product;
SELECT COUNT(*) FROM gold.dim_seller;
SELECT COUNT(*) FROM gold.dim_orders;
SELECT COUNT(*) FROM gold.dim_order_items;
SELECT COUNT(*) FROM gold.dim_payments;
```

```python
%sql
SELECT
    customer_id,
    COUNT(*) AS cnt
FROM gold.dim_customer
GROUP BY customer_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    product_id,
    COUNT(*) AS cnt
FROM gold.dim_product
GROUP BY product_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    seller_id,
    COUNT(*) AS row_count
FROM gold.dim_seller
GROUP BY seller_id
HAVING COUNT(*) > 1;
```
